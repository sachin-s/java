```python
import numpy as np
import pandas as pd
import random
import string
```

```python
"""
1. define name list and include values ranging from 10 to 100 with step value 10
2. create a dataframe multiplying each name to 10 times and reset the index
"""
names = [x for x in range(10,101,10)]
df = pd.DataFrame({'name':names*10}).sort_values('name').reset_index()
```

```python
# drop the index column
df.drop(columns=['index'],inplace=True)
```

```python
""" 
define the transform function palidromeAndNonPalindromeData(k) with the following functionalities
1. 

"""
def palidromeAndNonPalindromeData(k):
    i = k.iloc[0]
    l = len(k)
    a = [''.join(random.choices(string.ascii_letters,k=i)) for x in range(l//2)]
    b = [ y+y[::-1] for y in [''.join(random.choices(string.ascii_letters,k=i//2)) for x in range(l-(l//2))]] if i%2==0 else [ y+random.choice(string.ascii_letters)+y[::-1] for y in [''.join(random.choices(string.ascii_letters,k=(l-l//2))) for x in range(5)]]
    return a + b; 
```

```python
df['value'] = df.groupby('name')['name'].transform(palidromeAndNonPalindromeData)
```

```python
df['expected'] = df['value'].apply(lambda x: x[::-1])
```
